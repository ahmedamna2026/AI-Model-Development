## Section B: Functions

### B1. Why Functions Exist [Video]

**Task:** Fix Jo’s average; use one function called three times; honours cutoff: 85.

**Original Code**

```python
# Ali
ali = [72, 88, 95]
ali_avg = sum(ali) / len(ali)
if ali_avg >= 80:
    print("Ali: Honours", round(ali_avg, 1))
else:
    print("Ali: Pass", round(ali_avg, 1))
# Sam
sam = [64, 70, 81]
sam_avg = sum(sam) / len(sam)
if sam_avg >= 80:
    print("Sam: Honours", round(sam_avg, 1))
else:
    print("Sam: Pass", round(sam_avg, 1))
# Jo
jo = [90, 85, 99]
jo_avg = sum(sam) / len(jo)
if jo_avg >= 80:
    print("Jo: Honours", round(jo_avg, 1))
else:
    print("Jo: Pass", round(jo_avg, 1))
```

**Your Fix** #created functions with parameters


```python
def report_student(name, grades):
    average = sum(grades) / len(grades)
    if average >= 85:
        result = "Honours"
    else:
        result = "Pass"
    print(f"{name}: {result}", round(average, 1))

ali = [72, 88, 95]
sam = [64, 70, 81]
jo = [90, 85, 99]
report_student("Ali", ali)
report_student("Sam", sam)
report_student("Jo", jo)
```

### B3. Passing Objects Into Functions [Video]

**Task:** Change only the inside of add_bonus; preserve Alice’s original grades.

**Original Code**

```python
def add_bonus(grades): #alice's grades are passed into the function
    for i in range(len(grades)):
        grades[i] += 5
    return grades

def add_bonus_one(grade):
    grade += 5
    return grade

alice = [70, 80, 90] #grades stored in list
alice_curved = add_bonus(alice)
print("alice:       ", alice)
print("alice_curved:", alice_curved)
bob = 70 #grade is an integer. integer cannot be changed in place
bob_curved = add_bonus_one(bob)
print("bob:         ", bob)
print("bob_curved:  ", bob_curved)
```

## Section C: Classes

### C1. Why Classes Exist [Video]

**Task:** Use Student, add_grade, average, and report.

**Original Code**

```python
def make_student(name):
    return {"name": name, "grades": []}
def add_grade(student, grade):
    student["grades"].append(grade)
def average(student):
    return sum(student["grades"]) / len(student["grades"])
def report(student):
    print(f"{student['name']}: {average(student):.1f}")

ali = make_student("Ali")
add_grade(ali, 72)
add_grade(ali, 88)
report(ali)
```

**Your Fix** #class keeps data and actions together

```python
class Student:
    def __init__(self, name):#used _init_ to set up the student, self is the particular student 
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        return sum(self.grades) / len(self.grades)

    def report(self):
        print(f"{self.name}: {self.average():.1f}")

ali = Student("Ali")
ali.add_grade(72)
ali.add_grade(88)
ali.report()
```

### C3. Methods With and Without self [Video]

**Task:** Make both calls print A.

**Original Code**

```python
class Grader:
    def letter(score):#the original method expects one argument
        if score >= 80:
            return "A"
        elif score >= 65:
            return "B"
        return "C"

print(Grader.letter(85)) #calling through class works
g = Grader() #calling through object causes problem. Python passes g into an ordinary instance method.
print(g.letter(85)) #method receives 2 arguments object and score.
```

**Your Fix**

```python
class Grader:
    @staticmethod #tells python not to automatically pass the object
    def letter(score):
        if score >= 80:
            return "A"
        elif score >= 65:
            return "B"
        return "C"

print(Grader.letter(85))
g = Grader()
print(g.letter(85))
```

### C4. Edit In Place vs Return

**Task:** Return a new Student with a separate curved list; print both students’ grades.

**Original Code**

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades
    def curve(self, points):
        for i in range(len(self.grades)):
            self.grades[i] += points

ali = Student("Ali", [70, 80, 90])
ali = ali.curve(5)
print(ali.grades)
```

**Your Fix**

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def curve(self, points):
        curved_grades = []
        for grade in self.grades:
            curved_grades.append(grade + points)
        return Student(self.name, curved_grades)

ali = Student("Ali", [70, 80, 90])
ali_curved = ali.curve(5)
print("ali.grades:", ali.grades)
print("ali_curved.grades:", ali_curved.grades)
```
